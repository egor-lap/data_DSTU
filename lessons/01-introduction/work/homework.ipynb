## Подготовка

```python
from pathlib import Path
import sys
import pandas as pd

print("Python:", sys.executable, sys.version)
print("pandas:", pd.__version__)
print("Рабочая папка:", Path.cwd())
DATA = Path("../data")
assert DATA.is_dir(), "Рядом с папкой ноутбука должна находиться data/"
```

## Основной набор decisions.csv

```python
home_decisions = pd.read_csv(
    DATA / "decisions.csv", dtype={"lei": "string", "county_code": "string"}
)
home_outcomes = home_decisions["action_taken"]
home_n = len(home_decisions)
assert home_n > 0
assert home_outcomes.isna().sum() == 0
assert home_outcomes.isin([1, 2, 3]).sum() == home_n
print(home_decisions.shape)
home_outcomes.value_counts(dropna=False)
```

## H1. Доля отказов

```python
home_denied = home_outcomes.eq(3).sum()
home_denial_rate = home_denied / home_n
print(f"Отказы {home_denied}/{home_n}: {100 * home_denial_rate:.2f}%")
```

## H2. Одобрение без принятия предложения

```python
home_approved_not_accepted = home_outcomes.eq(2).sum()
home_approved_not_accepted_rate = home_approved_not_accepted / home_n
home_approval_rate = home_outcomes.isin([1, 2]).sum() / home_n
home_origination_rate = home_outcomes.eq(1).sum() / home_n
print(f"Одобрено, но не принято {home_approved_not_accepted}/{home_n}: "
      f"{100 * home_approved_not_accepted_rate:.2f}%")
print(f"Разность одобрения и выдачи: "
      f"{100 * (home_approval_rate - home_origination_rate):.2f} п.п.")
```

## H3. Объяснение инструментов
Series - проиндексированный одномерный массив значений
DataFrame - множество объектов Series
nrows = 3 - читает только первые три строки конкретно с файла
head(5) - читает первые 5 строк с уже созданной в памяти таблицы из исходного файла
Метод и атрибут: Метод - raw_outcomes.eq(1), Атрибут - home_decisions.shape Метод выполняет действие, атрибут возвращает уже хранящееся свойство объекта.
Ссылка на документацию и что удалось уточнить: https://pandas.pydata.org/docs/reference/api/pandas.Series.isin.html - isin принимает список значений и возвращает логическую Series; удобно для отбора исходов 1 и 2 одним условием.

## Исходный снимок для H+1

```python
raw = pd.read_csv(
    DATA / "hmda_2023_dc_raw.csv",
    dtype={"lei": "string", "county_code": "string"},
)
raw_outcomes = raw["action_taken"]
raw_n = len(raw)
assert raw_n > 0
print(raw.shape)
raw_outcomes.value_counts(dropna=False)
```

## H+1. Доли выдач в двух наборах

```python
raw_origination_rate = raw_outcomes.eq(1).sum() / raw_n
print(f"decisions: {100 * home_origination_rate:.2f}%")
print(f"raw: {100 * raw_origination_rate:.2f}%")
```
